##Step 3: Exploratory Data Analysis (EDA)

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

##Q1.Which Product Line Generates the Highest Total Sales?

```python
product_sales = supermarket.groupby("Product line")["Total"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
product_sales.plot(kind="bar")

plt.title("Total Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

###Analysis:
- Compare total sales across product lines.
- Identify the highest- and lowest-selling categories.
###Conclusion:

- The bar chart helps identify which product lines generate the highest and lowest sales, supporting better inventory and sales planning.

##Q2. How Do Total Sales Change Month by Month?

```python
supermarket["Date"] = pd.to_datetime(supermarket["Date"])

supermarket["Month"] = supermarket["Date"].dt.to_period("M")

monthly_sales = supermarket.groupby("Month")["Total"].sum()

plt.figure(figsize=(12, 5))
plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```

###Analysis:
- Observe changes in sales over time.
- Identify months with the highest and lowest sales.
###Conclusion:

- The line chart helps understand monthly sales trends and identify periods of high and low sales.

##Q3. What Is the Distribution of Different Payment Methods?

```python
payment_counts = supermarket["Payment"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Payment Method Distribution")
plt.show()
```

##Analysis:
- Identify the most frequently used payment method.
- Compare the percentage share of each payment method.
##Conclusion:

- The pie chart shows customer preferences for different payment methods and helps identify the most commonly used payment option.

##Q4. What Is the Distribution of Product Unit Prices?

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=supermarket,
    x="Unit price",
    bins=20,
    kde=True
)

plt.title("Distribution of Unit Prices")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

##Analysis:
- Identify the most common product price ranges.
- Observe the spread and shape of the price distribution.
##Conclusion:

- The histogram shows the distribution of product prices and helps understand the supermarket's pricing patterns.

##Q5. Is There a Relationship Between Quantity Purchased and Total Sales?

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=supermarket,
    x="Quantity",
    y="Total"
)

plt.title("Quantity Purchased vs Total Sales")
plt.xlabel("Quantity")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()

correlation = supermarket["Quantity"].corr(supermarket["Total"])

print("Correlation:", round(correlation, 3))
```

##Analysis:
- Observe how total sales vary with quantity purchased.
- Check whether higher quantities are generally associated with higher transaction values.
- Examine the correlation coefficient.
##Conclusion:

- The scatter plot helps analyze the relationship between quantity purchased and total sales. The correlation coefficient indicates the strength and direction of their linear relationship.

##Q6. How Does Gross Income Vary Across Different Product Lines?

```python
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=supermarket,
    x="Product line",
    y="gross income"
)

plt.title("Gross Income Distribution by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Gross Income")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Summary statistics
income_summary = supermarket.groupby("Product line")["gross income"].agg(
    ["mean", "median", "min", "max"]
)
```

##Analysis:
- Compare median gross income across product lines.
- Identify categories with greater variation in gross income.
- Observe potential outliers.
##Conclusion:

- The box plot compares gross income distributions across product lines and helps identify differences in typical income and transaction-level variability.
